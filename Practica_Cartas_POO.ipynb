**Ejercicio 6**: Escribir un método en la clase Mazo que *mezcle* el mazo. Puede ser de utilidad el módulo [`random`](https://docs.python.org/3/library/random.html) de la biblioteca estándar de Python.

```python
import random


class Carta:

    apodos = {
        1: "As",
        10: "Sota",
        11: "Caballo",
        12: "Rey"
    }

    def __init__(self, numero, palo):

        self.numero = numero
        self.palo = palo

    def __str__(self):

        if self.numero in Carta.apodos:
            nombre = Carta.apodos[self.numero]
        else:
            nombre = str(self.numero)

        return f"{nombre} de {self.palo}"


class Mazo:

    def __init__(self):

        self.cartas = []

        palos = ["Espadas", "Bastos", "Copas", "Oros"]

        for palo in palos:

            for numero in range(1, 13):

                if numero != 8 and numero != 9:

                    carta = Carta(numero, palo)
                    self.cartas.append(carta)

    def cantidad(self):

        return len(self.cartas)

    def mezclar(self):

        random.shuffle(self.cartas)

    def sacar_carta(self):

        carta_sacada = self.cartas.pop()

        return str(carta_sacada)


baraja = Mazo()

baraja.mezclar()

print(baraja.sacar_carta())

print("Cantidad de cartas en la baraja:", baraja.cantidad())
    
```
